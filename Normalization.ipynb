# Data Normalization

```python
import pandas as pd
```

```python
# paths
in_path  = "/mnt/data/mouse_events_0601.csv"
out_path = "/mnt/data/mouse_events_0601_norm.csv"

# grid size
GRID_W, GRID_H = 8, 6

# load data
df = pd.read_csv(in_path)

# ensure numeric 
for col in ["x", "y", "page_w", "page_h", "t_ms"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# keep only valid rows
mask = (
    df["x"].notna() & df["y"].notna() &
    df["page_w"].notna() & df["page_h"].notna() &
    (df["page_w"] > 0) & (df["page_h"] > 0)
)
df = df.loc[mask].copy()

# normalization to 8x6 grid 
# x_norm in [0,8], y_norm in [0,6]
df["x_norm_8"] = df["x"] * GRID_W / df["page_w"]
df["y_norm_6"] = df["y"] * GRID_H / df["page_h"]

# save
df.to_csv(out_path, index=False)

print("Saved:", out_path)
print(df[["x", "y", "page_w", "page_h", "x_norm_8", "y_norm_6"]].head())
```
